```python
# Step 1: Load the data
df = pd.read_csv("dataset/Copy of ai_job_dataset.csv")

# Step 2: Split into a list (comma-separated)
df['skill_list'] = df['required_skills'].apply(lambda x: [i.strip().title() for i in str(x).split(',') if i.strip()])

# Step 3: Explode into multiple rows
df_exploded = df.explode('skill_list')

# Step 4: Rename for clarity
df_exploded = df_exploded.rename(columns={'skill_list': 'skill'})

# Step 5: Drop blank/null skill rows
df_exploded = df_exploded[df_exploded['skill'].notnull()]
df_exploded = df_exploded[df_exploded['skill'] != '']


# Step 6: Export to CSV
df_exploded.to_csv("exploded_skills_clean.csv", index=False)
```
